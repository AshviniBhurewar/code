```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("sales_data_100_rows (2).csv")

sales = df.groupby("Product")["Sales_Amount"].sum()

plt.bar(sales.index, sales.values)
plt.title("Sales by Product")
plt.xlabel("Product")
plt.ylabel("Sales")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("sales_data_100_rows (2).csv")

region = df.groupby("Region")["Sales_Amount"].sum()

plt.pie(region.values, labels=region.index,
        autopct="%1.1f%%")
plt.title("Sales by Region")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("sales_data_100_rows (2).csv")

sales = df.groupby("Product")["Sales_Amount"].sum()

plt.plot(sales.index, sales.values, marker="o")
plt.title("Product Sales")
plt.xlabel("Product")
plt.ylabel("Sales")
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("sales_data_100_rows (2).csv")

sns.boxplot(x="Region", y="Sales_Amount", data=df)
plt.title("Sales Distribution by Region")
plt.show()
```
